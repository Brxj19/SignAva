```python
import io
import pickle
import torch
import numpy as np
from pathlib import Path


class CPU_Unpickler(pickle.Unpickler):
    def find_class(self, module, name):
        if module == "torch.storage" and name == "_load_from_bytes":
            return lambda b: torch.load(
                io.BytesIO(b),
                map_location=torch.device("cpu"),
                weights_only=False
            )
        return super().find_class(module, name)


def load_pickle_cpu(path):
    path = Path(path)

    # First try CPU-safe pickle loading
    try:
        with open(path, "rb") as f:
            return CPU_Unpickler(f).load()
    except Exception as e1:
        print("CPU_Unpickler failed:", e1)

    # Fallback: try torch.load directly
    try:
        return torch.load(path, map_location=torch.device("cpu"), weights_only=False)
    except Exception as e2:
        print("torch.load fallback failed:", e2)
        raise e2
```

```python
# ============================================================
# Render real dataset SMPL-X animations for selected 10 glosses
# Uses CUDA for renderer/SMPL-X forward pass when available
# ============================================================

from pathlib import Path
import os
import sys
import json
import pickle
import io
import re
import inspect
import numpy as np
import torch

# ------------------------------------------------------------
# 1. Project setup
# ------------------------------------------------------------
PROJECT_ROOT = Path("/content/drive/MyDrive/FYP/SignAvatar_Generative_Project")
os.chdir(PROJECT_ROOT)

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
print("Current directory:", Path.cwd())

# ------------------------------------------------------------
# 2. CUDA setup
# ------------------------------------------------------------
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

print("\nDevice selected:", DEVICE)

if DEVICE == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("CUDA not available. Rendering will use CPU.")

# ------------------------------------------------------------
# 3. Target glosses
# ------------------------------------------------------------
TARGET_GLOSSES = [
    "a",
    "a lot",
    "abdomen",
    "able",
    "about",
    "above",
    "accent",
    "accept",
    "accident",
    "accomplish",
]

# Choose which dataset sample to render for each gloss:
# "first"   -> first sample from index
# "longest" -> sample with maximum frame length, usually better for visible motion
SAMPLE_STRATEGY = "longest"

# Optional: limit frames for faster testing.
# Set to None for full real sequence.
MAX_FRAMES = None

# ------------------------------------------------------------
# 4. CPU-safe pickle loader
# ------------------------------------------------------------
class CPU_Unpickler(pickle.Unpickler):
    def find_class(self, module, name):
        if module == "torch.storage" and name == "_load_from_bytes":
            return lambda b: torch.load(
                io.BytesIO(b),
                map_location=torch.device("cpu"),
                weights_only=False
            )
        return super().find_class(module, name)

def load_pickle_cpu(path):
    with open(path, "rb") as f:
        return CPU_Unpickler(f).load()

# ------------------------------------------------------------
# 5. Helpers
# ------------------------------------------------------------
def safe_name(text):
    text = text.lower().strip().replace(" ", "_")
    text = re.sub(r"[^a-z0-9_]+", "", text)
    return text

def resolve_path(path_like):
    """
    Resolve paths from index.json.
    Supports:
    - relative paths from project root
    - absolute paths
    """
    p = Path(path_like)

    if p.exists():
        return p

    p2 = PROJECT_ROOT / path_like
    if p2.exists():
        return p2

    p3 = PROJECT_ROOT / str(path_like).lstrip("/")
    if p3.exists():
        return p3

    return p2

def extract_smplx_motion(pkl_path):
    data = load_pickle_cpu(pkl_path)

    if "smplx" not in data:
        raise KeyError(f"'smplx' key not found in {pkl_path}")

    motion = np.asarray(data["smplx"], dtype=np.float32)

    if motion.ndim != 2:
        raise ValueError(f"Expected 2D SMPL-X motion, got shape {motion.shape}")

    if motion.shape[1] != 182:
        raise ValueError(f"Expected motion shape (T, 182), got {motion.shape}")

    if MAX_FRAMES is not None:
        motion = motion[:MAX_FRAMES]

    return motion

def choose_sample_for_gloss(sample_paths, strategy="longest"):
    """
    Select one real dataset sample for a gloss.
    """
    if len(sample_paths) == 0:
        raise ValueError("No sample paths available.")

    if strategy == "first":
        p = resolve_path(sample_paths[0])
        motion = extract_smplx_motion(p)
        return p, motion

    if strategy == "longest":
        best_path = None
        best_motion = None
        best_len = -1

        for rel_path in sample_paths:
            p = resolve_path(rel_path)

            if not p.exists():
                continue

            try:
                motion = extract_smplx_motion(p)
                seq_len = motion.shape[0]

                if seq_len > best_len:
                    best_len = seq_len
                    best_path = p
                    best_motion = motion

            except Exception as e:
                print("Skipping bad sample:", p, "|", e)

        if best_path is None:
            raise FileNotFoundError("No valid PKL sample found for this gloss.")

        return best_path, best_motion

    raise ValueError(f"Unknown SAMPLE_STRATEGY: {strategy}")

# ------------------------------------------------------------
# 6. Locate index.json
# ------------------------------------------------------------
possible_index_paths = [
    PROJECT_ROOT / "data/processed/index.json",
    PROJECT_ROOT / "data/index.json",
]

index_path = None

for p in possible_index_paths:
    if p.exists():
        index_path = p
        break

if index_path is None:
    raise FileNotFoundError(
        "Could not find index.json. Checked data/processed/index.json and data/index.json"
    )

print("\nUsing index:", index_path)

with open(index_path, "r") as f:
    index = json.load(f)

if "gloss_to_paths" not in index:
    raise KeyError("index.json does not contain 'gloss_to_paths'")

gloss_to_paths = index["gloss_to_paths"]

# ------------------------------------------------------------
# 7. Output directories
# ------------------------------------------------------------
reference_motion_dir = PROJECT_ROOT / "outputs/dataset_reference"
reference_video_dir = PROJECT_ROOT / "outputs/videos/dataset_reference"

reference_motion_dir.mkdir(parents=True, exist_ok=True)
reference_video_dir.mkdir(parents=True, exist_ok=True)

print("\nReference motion dir:", reference_motion_dir)
print("Reference video dir:", reference_video_dir)

# ------------------------------------------------------------
# 8. Import renderer and force CUDA if supported
# ------------------------------------------------------------
from src.renderer import smplx_renderer as renderer_module

print("\nRenderer module loaded.")

def render_motion_with_cuda(motion_path, output_path):
    """
    Render a motion .npy using CUDA if the renderer supports device argument.
    Falls back to SMPLXRenderer class if needed.
    """

    motion_path = Path(motion_path)
    output_path = Path(output_path)

    # Try top-level render_generated_motion() first
    if hasattr(renderer_module, "render_generated_motion"):
        func = renderer_module.render_generated_motion
        sig = inspect.signature(func)

        kwargs = {
            "motion_path": motion_path,
            "output_path": output_path,
        }

        if "device" in sig.parameters:
            kwargs["device"] = DEVICE

        if "max_frames" in sig.parameters and MAX_FRAMES is not None:
            kwargs["max_frames"] = MAX_FRAMES

        try:
            return func(**kwargs)
        except TypeError as e:
            print("Top-level render function TypeError, trying class renderer:", e)

    # Try class-based renderer
    if hasattr(renderer_module, "SMPLXRenderer"):
        RendererClass = renderer_module.SMPLXRenderer
        renderer_sig = inspect.signature(RendererClass)

        init_kwargs = {}

        if "device" in renderer_sig.parameters:
            init_kwargs["device"] = DEVICE

        renderer = RendererClass(**init_kwargs)

        method = renderer.render_generated_motion
        method_sig = inspect.signature(method)

        method_kwargs = {
            "motion_path": motion_path,
            "output_path": output_path,
        }

        if "max_frames" in method_sig.parameters and MAX_FRAMES is not None:
            method_kwargs["max_frames"] = MAX_FRAMES

        return method(**method_kwargs)

    raise RuntimeError(
        "Could not find render_generated_motion() or SMPLXRenderer in src.renderer.smplx_renderer"
    )

# ------------------------------------------------------------
# 9. Render real dataset reference videos
# ------------------------------------------------------------
results = []

print("\n====================================================")
print("Rendering real dataset animations")
print("Sample strategy:", SAMPLE_STRATEGY)
print("Renderer device:", DEVICE)
print("====================================================")

for gloss in TARGET_GLOSSES:
    print("\n----------------------------------------------------")
    print("Gloss:", gloss)

    safe = safe_name(gloss)

    try:
        if gloss not in gloss_to_paths:
            raise KeyError(f"Gloss '{gloss}' not found in index.json")

        sample_paths = gloss_to_paths[gloss]
        print("Available samples:", len(sample_paths))

        pkl_path, motion = choose_sample_for_gloss(
            sample_paths,
            strategy=SAMPLE_STRATEGY
        )

        print("Selected PKL:", pkl_path)
        print("Real motion shape:", motion.shape)

        motion_path = reference_motion_dir / f"{safe}_real_smplx.npy"
        video_path = reference_video_dir / f"{safe}_real_reference.mp4"

        np.save(motion_path, motion)
        print("Saved real motion:", motion_path)

        rendered_path = render_motion_with_cuda(
            motion_path=motion_path,
            output_path=video_path
        )

        print("Saved video:", rendered_path)

        results.append({
            "gloss": gloss,
            "safe_gloss": safe,
            "status": "done",
            "sample_strategy": SAMPLE_STRATEGY,
            "pkl_path": str(pkl_path),
            "motion_shape": list(motion.shape),
            "motion_path": str(motion_path),
            "video_path": str(rendered_path),
            "device": DEVICE,
            "error": None,
        })

    except Exception as e:
        print("FAILED:", gloss, "|", str(e))

        results.append({
            "gloss": gloss,
            "safe_gloss": safe,
            "status": "failed",
            "sample_strategy": SAMPLE_STRATEGY,
            "pkl_path": None,
            "motion_shape": None,
            "motion_path": None,
            "video_path": None,
            "device": DEVICE,
            "error": str(e),
        })

# ------------------------------------------------------------
# 10. Save reference index
# ------------------------------------------------------------
reference_index_path = reference_motion_dir / "dataset_reference_index.json"

with open(reference_index_path, "w") as f:
    json.dump(results, f, indent=2)

print("\n====================================================")
print("Dataset reference rendering complete.")
print("Reference index:", reference_index_path)
print("====================================================")

for item in results:
    print(item["gloss"], "→", item["status"], "→", item.get("video_path"))

# ------------------------------------------------------------
# 11. Display videos inside Colab
# ------------------------------------------------------------
try:
    from IPython.display import Video, display

    print("\nDisplaying generated dataset reference videos:")

    for item in results:
        if item["status"] == "done":
            print("\nReal dataset reference:", item["gloss"])
            display(Video(item["video_path"], embed=True))
        else:
            print("\nFailed:", item["gloss"], "|", item["error"])

except Exception as e:
    print("Could not display videos inline:", e)
```
